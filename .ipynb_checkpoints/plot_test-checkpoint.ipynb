## Report figure

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

plt.rcParams['font.size'] = 12
plt.rcParams["font.family"] = "Arial"

# Parameters
num_ticks = 3
tick_positions = list(range(num_ticks))
rect_width = 0.5
rect_height = 0.3

fig, ax = plt.subplots(figsize=(8, 3), dpi=200)

# Draw rectangles above each tick
for x in tick_positions:
    rect = patches.Rectangle(
        (x - rect_width/2, 0), rect_width, rect_height,
        linewidth=1, edgecolor='skyblue', facecolor='skyblue', alpha=0.5
    )
    ax.add_patch(rect)

# Averaging interval line (above first rectangle)
ax.annotate(
    "", xy=(tick_positions[0] - rect_width/2 - arrow_offset, 0.4), xytext=(tick_positions[0] + rect_width/2 + arrow_offset, 0.4),
    arrowprops=dict(arrowstyle='<->')
)
ax.text(tick_positions[0], 0.5, "Averaging interval", ha='center', va='bottom')

# Measurement interval line (below x axis)
ax.annotate(
    "", xy=(tick_positions[0]-arrow_offset, -0.1), xytext=(tick_positions[1]+arrow_offset, -0.1),
    arrowprops=dict(arrowstyle='<->')
)
ax.text((tick_positions[0] + tick_positions[1]) / 2, -0.2, "Measurement interval", ha='center', va='top')

# Hide all spines
for spine in ax.spines.values():
    spine.set_visible(False)

# Draw custom x-axis as an arrow from leftmost to rightmost point
arrow_start = tick_positions[0] - 0.5
arrow_end = tick_positions[-1] + 0.6
ax.annotate(
    "", xy=(arrow_end, 0), xytext=(arrow_start, 0),
    arrowprops=dict(arrowstyle='->', linewidth=1.5)
)
ax.text(arrow_end + 0.03, 0, "Time", va='center')

# Manual x-ticks at y=0
tick_length = 0.1
for x in tick_positions:
    ax.plot([x, x], [-tick_length/2, tick_length/2], color='black', lw=1)

# Set ticks and limits
ax.set_xticks([])
ax.set_yticks([])
ax.set_xlim(arrow_start - 0.5, arrow_end + 0.5)
ax.set_ylim(-1, 1)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create a 2D array using sine functions
x = np.linspace(0, 2 * np.pi, 30)
arr = np.sin(x+0.7)

# Averaging function
def average_pixels(arr, xpixavg=1, ypixavg=1):
    is_1d = arr.ndim == 1
    arr = np.atleast_2d(arr)

    ny, nx = arr.shape
    ny_crop = ny - ny % ypixavg
    nx_crop = nx - nx % xpixavg

    arr_cropped = arr[:ny_crop, :nx_crop]
    reshaped = arr_cropped.reshape(ny_crop // ypixavg, ypixavg, nx_crop // xpixavg, xpixavg)
    block_avg = reshaped.mean(axis=(1, 3))
    upscaled = np.kron(block_avg, np.ones((ypixavg, xpixavg)))

    # Pad to original size
    result = np.full_like(arr, np.nan, dtype=float)
    result[:upscaled.shape[0], :upscaled.shape[1]] = upscaled

    return result.squeeze() if is_1d else result


# Apply averaging
arr_avg = average_pixels(arr, xpixavg=3, ypixavg=1)

# Plot original and averaged arrays
fig, ax = plt.subplots(figsize=(10, 4), dpi=150)
ax.plot(x, arr, marker='o', lw=3)
ax.scatter(x, arr_avg, c='tab:orange', s=2,zorder=3)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

ds = ds_disk
PT_colors = ['tab:blue', 'tab:orange']

fig = plt.figure(figsize=(8, 2), dpi=150)
ax1 = fig.add_subplot()
ax2 = ax1.twinx()

# Plot pressure
ax1.plot(ds['TIME'], ds['P'], color=PT_colors[0])

# Plot temperature
ax2.plot(ds['TIME'], ds['T'], color=PT_colors[1])

ax1.set_ylabel('Pressure (dBar)', color=PT_colors[0])
ax1.tick_params(which='both', axis='y', colors=PT_colors[0])

ax2.set_ylabel('Temperature (°C)', color=PT_colors[1])
ax2.tick_params(which='both', axis='y', colors=PT_colors[1])

ax1.minorticks_on()
ax2.minorticks_on()
```
